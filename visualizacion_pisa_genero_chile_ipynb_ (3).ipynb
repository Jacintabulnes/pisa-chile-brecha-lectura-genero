```python

import pandas as pd

# Puntajes transcritos del gráfico oficial PISA 2022,
# página 26, Agencia de Calidad de la Educación.
puntajes = pd.DataFrame({
    "anio": [2000, 2006, 2009, 2012, 2015, 2018, 2022],
    "Mujeres": [421, 451, 461, 452, 465, 462, 451],
    "Hombres": [396, 434, 439, 430, 453, 442, 445]
})

df_csv = puntajes.melt(
    id_vars="anio",
    var_name="sexo",
    value_name="puntaje_lectura"
)

df_csv["fuente"] = "Agencia de Calidad de la Educación, PISA 2022"

df_csv.to_csv(
    "pisa_chile_lectura_genero_2000_2022.csv",
    index=False
)

print("CSV preparado correctamente")
```

```python
import pandas as pd

df = pd.read_csv("pisa_chile_lectura_genero_2000_2022.csv")

display(df)
```

```python

import altair as alt

grafico = (
    alt.Chart(df)
    .mark_line(point=True, strokeWidth=3)
    .encode(
        x=alt.X(
            "anio:O",
            title="Año de evaluación"
        ),
        y=alt.Y(
            "puntaje_lectura:Q",
            title="Puntaje promedio en lectura",
            scale=alt.Scale(domain=[380, 480])
        ),
        color=alt.Color(
            "sexo:N",
            title="Sexo",
            scale=alt.Scale(
                domain=["Mujeres", "Hombres"],
                range=["#008F83", "#E16B55"]
            )
        ),
        tooltip=[
            alt.Tooltip("anio:O", title="Año"),
            alt.Tooltip("sexo:N", title="Sexo"),
            alt.Tooltip("puntaje_lectura:Q", title="Puntaje")
        ]
    )
    .properties(
        width=700,
        height=380,
        title="La ventaja lectora de las mujeres chilenas se estrecha"
    )
)

grafico
```

```python

import altair as alt
import pandas as pd

# Preparar los datos
datos = df.copy()
datos["anio"] = datos["anio"].astype(int)
datos["puntaje_lectura"] = pd.to_numeric(
    datos["puntaje_lectura"]
)

colores = alt.Scale(
    domain=["Mujeres", "Hombres"],
    range=["#008F83", "#E16B55"]
)

base = alt.Chart(datos).encode(
    x=alt.X(
        "anio:O",
        title="Año de evaluación",
        axis=alt.Axis(labelAngle=0)
    ),
    y=alt.Y(
        "puntaje_lectura:Q",
        title="Puntaje promedio en lectura",
        scale=alt.Scale(domain=[380, 485])
    ),
    color=alt.Color(
        "sexo:N",
        scale=colores,
        title="Sexo"
    ),
    tooltip=[
        alt.Tooltip("anio:O", title="Año"),
        alt.Tooltip("sexo:N", title="Sexo"),
        alt.Tooltip(
            "puntaje_lectura:Q",
            title="Puntaje"
        )
    ]
)

lineas = base.mark_line(strokeWidth=3)

puntos = base.mark_circle(size=85)

etiquetas = (
    alt.Chart(datos[datos["anio"] == 2022])
    .mark_text(
        align="left",
        dx=10,
        fontSize=14,
        fontWeight="bold"
    )
    .encode(
        x="anio:O",
        y=alt.Y(
            "puntaje_lectura:Q",
            scale=alt.Scale(domain=[380, 485])
        ),
        text="puntaje_lectura:Q",
        color=alt.Color(
            "sexo:N",
            scale=colores,
            legend=None
        )
    )
)

grafico_final = (
    (lineas + puntos + etiquetas)
    .properties(
        width=750,
        height=400,
        title=alt.TitleParams(
            text="La ventaja lectora de las mujeres chilenas se estrecha",
            subtitle=[
                "Entre 2018 y 2022, la brecha pasó de 20 a 6 puntos.",
                "En 2022, la diferencia dejó de ser estadísticamente significativa."
            ],
            fontSize=21,
            subtitleFontSize=13,
            anchor="start"
        )
    )
    .configure_view(stroke=None)
    .configure_axis(
        gridColor="#EAEAEA",
        labelFontSize=12,
        titleFontSize=13
    )
    .configure_legend(
        labelFontSize=12,
        titleFontSize=12,
        orient="bottom"
    )
)

grafico_final
```
